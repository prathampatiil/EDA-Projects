**Task 4. Data Cleaning**

```python
data.isnull().sum()

#dropping all missing values from the dataset
data.dropna(inplace=True)

#data.fillna
data.isnull().sum()
```

```python
data.shape
```

```python
# dealing with duplicates rows
data.duplicated().sum()

data.drop_duplicates(inplace=True)
data.duplicated().sum()
```

```python
data.dtypes
data[['id', 'name', 'host_id','host_name','neighbourhood_group', 'neighbourhood', 'room_type', 'last_review', 'license', 'rating', 'bedrooms', 'baths']] = data[['id', 'name', 'host_id','host_name', 'neighbourhood_group', 'neighbourhood', 'room_type', 'last_review', 'license', 'rating', 'bedrooms', 'baths']].astype('object')
data.dtypes
```

### EDA
**Task 5.Data Analysis**

**Univariate Analysis**

```python
#identifying outilers in price
df=data[data['price']<1500]
sns.boxplot(data=df, x='price')
```

```python
#Price disribution
plt.Figure(figsize=(8,5))
sns.histplot(data=df, x='price', bins=50, kde=True)
plt.title('Price Distribution')
plt.ylabel("Frequecy")
plt.show()
```

```python
df.columns
```

```python
df.groupby('neighbourhood_group')['price per bed'].mean()
```

```python
df.groupby('neighbourhood_group')['price'].mean()
```

Feature Engineering

```python
df['price per bed']=df['price']/df['beds']
df.head()
```

Bi Variate Analysis

```python
df.columns
```

```python
sns.barplot(data=df,x='neighbourhood_group',y='price',hue='room_type')
```

```python
#number of reviews and price
sns.scatterplot(data=df,x='number_of_reviews',y='price',hue='neighbourhood_group')
```

```python
df.dtypes
```

```python
sns.pairplot(data=df,vars=['price','minimum_nights','number_of_reviews','reviews_per_month','availability_365'],hue='neighbourhood_group')
```

```python
#geographical distribution of listings
sns.scatterplot(data=df,x='longitude',y='latitude',hue='neighbourhood_group')
plt.title('Geographical disrtibution of Air BnB Listings')
plt.show()
```

```python
#heatmap of correlation between numerical features
plt.figure(figsize=(10,6))
corr = df[['latitude', 'longitude', 'price', 'minimum_nights', 'number_of_reviews', 'reviews_per_month', 'calculated_host_listings_count', 'availability_365','beds']].corr()

sns.heatmap(corr, annot=True, cmap='coolwarm')
```
